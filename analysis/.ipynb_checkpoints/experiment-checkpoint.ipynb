### 1.2 Reliability Analysis

```python
def cronbach_alpha(data):
    """
    Calculate Cronbach's alpha.

    Rows:
        Participant-video evaluations

    Columns:
        Kansei items
    """

    clean_data = data.dropna()

    number_of_items = clean_data.shape[1]

    if number_of_items < 2:
        return np.nan

    item_variances = clean_data.var(
        axis=0,
        ddof=1
    )

    total_score = clean_data.sum(axis=1)

    total_variance = total_score.var(ddof=1)

    if total_variance == 0 or pd.isna(total_variance):
        return np.nan

    alpha = (
        number_of_items /
        (number_of_items - 1)
    ) * (
        1 -
        item_variances.sum() /
        total_variance
    )

    return alpha
```

```python
def interpret_alpha(alpha):
    if pd.isna(alpha):
        return "Cannot calculate"
    elif alpha >= 0.90:
        return "Excellent"
    elif alpha >= 0.80:
        return "Good"
    elif alpha >= 0.70:
        return "Acceptable"
    elif alpha >= 0.60:
        return "Questionable"
    elif alpha >= 0.50:
        return "Poor"
    else:
        return "Unacceptable"
```

```python
reliability_results = []

for dimension, columns in kansei_dimensions.items():

    alpha = cronbach_alpha(df[columns])

    reliability_results.append({
        "Dimension": dimension,
        "Number of Items": len(columns),
        "Complete Evaluations": df[columns].dropna().shape[0],
        "Cronbach Alpha": alpha,
        "Interpretation": interpret_alpha(alpha)
    })

reliability_results = pd.DataFrame(
    reliability_results
)

display(reliability_results.round(3))
```

```python
overall_alpha = cronbach_alpha(
    df[kansei_columns]
)

overall_reliability = pd.DataFrame({
    "Scale": ["All Kansei Words"],
    "Number of Items": [len(kansei_columns)],
    "Complete Evaluations": [
        df[kansei_columns].dropna().shape[0]
    ],
    "Cronbach Alpha": [overall_alpha],
    "Interpretation": [
        interpret_alpha(overall_alpha)
    ]
})

display(overall_reliability.round(3))
```

```python
def item_reliability_analysis(data):
    """
    Calculate corrected item-total correlation
    and Cronbach's alpha if each item is deleted.
    """

    clean_data = data.dropna()

    results = []

    for column in clean_data.columns:

        remaining_columns = [
            other_column
            for other_column in clean_data.columns
            if other_column != column
        ]

        other_item_total = clean_data[
            remaining_columns
        ].sum(axis=1)

        item_total_correlation = clean_data[
            column
        ].corr(other_item_total)

        alpha_if_deleted = cronbach_alpha(
            clean_data[remaining_columns]
        )

        results.append({
            "Column": column,
            "Kansei Word": kansei_labels[column],
            "Corrected Item-Total Correlation":
                item_total_correlation,
            "Cronbach Alpha if Deleted":
                alpha_if_deleted
        })

    return pd.DataFrame(results)
```

```python
item_reliability_tables = {}

for dimension, columns in kansei_dimensions.items():

    print(f"\n{dimension} reliability details")

    item_table = item_reliability_analysis(
        df[columns]
    )

    item_reliability_tables[dimension] = item_table

    display(item_table.round(3))
```

### 1.3 Kansei Word Weight Calculation

```python
weight_results = []

for dimension, columns in kansei_dimensions.items():

    standard_deviations = df[columns].std(
        ddof=1
    )

    sd_total = standard_deviations.sum()

    if sd_total == 0 or pd.isna(sd_total):
        # Equal weights if all standard deviations are zero
        normalized_weights = pd.Series(
            1 / len(columns),
            index=columns
        )
    else:
        normalized_weights = (
            standard_deviations / sd_total
        )

    for column in columns:
        weight_results.append({
            "Dimension": dimension,
            "Column": column,
            "Kansei Word": kansei_labels[column],
            "Mean": df[column].mean(),
            "SD": standard_deviations[column],
            "Within-Dimension Weight":
                normalized_weights[column]
        })

weight_table = pd.DataFrame(weight_results)

display(weight_table.round(4))
```

```python
weight_check = (
    weight_table
    .groupby("Dimension")[
        "Within-Dimension Weight"
    ]
    .sum()
    .reset_index(name="Weight Sum")
)

display(weight_check.round(6))
```

### 1.4 Kansei Score Calculation

```python
def weighted_score_with_missing(data, columns, weights):
    """
    Calculate a weighted Kansei Score while handling missing values.

    Parameters
    ----------
    data : pandas.DataFrame
        Dataset containing the Kansei ratings.

    columns : list
        Kansei columns included in the score.

    weights : dict
        Dictionary containing the weight of each Kansei word.

    Returns
    -------
    pandas.Series
        Weighted scores on the -3 to 3 scale.
    """

    # Convert the weight dictionary into a Series
    weight_series = pd.Series(weights).reindex(columns)

    # Multiply each Kansei rating by its weight
    weighted_values = data[columns].mul(
        weight_series,
        axis=1
    )

    # Calculate the total available weight for each row
    available_weights = (
        data[columns]
        .notna()
        .mul(weight_series, axis=1)
        .sum(axis=1)
    )

    # Calculate the weighted sum
    weighted_sum = weighted_values.sum(
        axis=1,
        min_count=1
    )

    # Divide by the available weight
    weighted_score = (
        weighted_sum /
        available_weights.replace(0, np.nan)
    )

    return weighted_score
```

```python
for dimension, columns in kansei_dimensions.items():

    score_column = f"{dimension}_Kansei_Score"

    df[score_column] = weighted_score_with_missing(
        data=df,
        columns=columns,
        weights=within_dimension_weights[dimension]
    )

dimension_score_columns = [
    f"{dimension}_Kansei_Score"
    for dimension in kansei_dimensions
]

display(
    df[
        [
            "participant_id",
            "group",
            "video_id",
            "video_id_count"
        ] +
        dimension_score_columns
    ].head(24).round(3)
)
```

### 2.1 Overall Kansei Score by Topic

```python
def confidence_interval_95(series):
    """
    Return the lower and upper limits of the
    95% confidence interval for a numeric series.
    """

    clean = series.dropna()
    n = len(clean)

    if n < 2:
        return pd.Series({
            "CI_Lower": np.nan,
            "CI_Upper": np.nan
        })

    mean = clean.mean()
    standard_error = stats.sem(clean)

    margin = stats.t.ppf(
        0.975,
        df=n - 1
    ) * standard_error

    return pd.Series({
        "CI_Lower": mean - margin,
        "CI_Upper": mean + margin
    })
```

```python
overall_topic_summary = (
    df.groupby("Topic", observed=False)[
        "Overall_Kansei_Score"
    ]
    .agg(
        Mean="mean",
        SD="std",
        Median="median",
        Minimum="min",
        Maximum="max",
        N="count"
    )
    .reset_index()
)

overall_ci = (
    df.groupby("Topic", observed=False)[
        "Overall_Kansei_Score"
    ]
    .apply(confidence_interval_95)
    .unstack()
    .reset_index()
)

overall_topic_summary = overall_topic_summary.merge(
    overall_ci,
    on="Topic",
    how="left"
)

display(overall_topic_summary.round(3))
```

### 2.4 Statistical Tests

```python
overall_wide = (
    participant_topic_scores
    .pivot(
        index="participant_id",
        columns="Topic",
        values="Overall_Kansei_Score"
    )
    .reindex(columns=topic_order)
    .dropna()
)

display(overall_wide.head().round(3))
```

```python
normality_results = []

for topic in topic_order:
    values = overall_wide[topic].dropna()

    if len(values) >= 3:
        statistic, p_value = stats.shapiro(
            values
        )
    else:
        statistic = np.nan
        p_value = np.nan

    normality_results.append({
        "Topic": topic,
        "N": len(values),
        "Shapiro W": statistic,
        "p-value": p_value,
        "Normal at 0.05":
            p_value > 0.05
            if not pd.isna(p_value)
            else np.nan
    })

normality_results = pd.DataFrame(
    normality_results
)

display(normality_results.round(4))
```

```python
anova_overall_data = (
    participant_topic_scores[
        [
            "participant_id",
            "Topic",
            "Overall_Kansei_Score"
        ]
    ]
    .dropna()
)

overall_rm_anova = AnovaRM(
    data=anova_overall_data,
    depvar="Overall_Kansei_Score",
    subject="participant_id",
    within=["Topic"]
).fit()

print(overall_rm_anova)
```

```python
friedman_statistic, friedman_p = (
    stats.friedmanchisquare(
        overall_wide["Language"],
        overall_wide["Science"],
        overall_wide["History"]
    )
)

friedman_overall_result = pd.DataFrame({
    "Test": ["Friedman"],
    "Statistic": [friedman_statistic],
    "p-value": [friedman_p],
    "Significant": [friedman_p < 0.05]
})

display(friedman_overall_result.round(4))
```

```python
pairwise_overall_results = []

topic_pairs = list(
    combinations(topic_order, 2)
)

for topic_1, topic_2 in topic_pairs:

    paired_data = overall_wide[
        [topic_1, topic_2]
    ].dropna()

    t_statistic, raw_p = stats.ttest_rel(
        paired_data[topic_1],
        paired_data[topic_2]
    )

    difference = (
        paired_data[topic_1] -
        paired_data[topic_2]
    )

    mean_difference = difference.mean()

    difference_sd = difference.std(ddof=1)

    if difference_sd == 0:
        cohens_dz = np.nan
    else:
        cohens_dz = (
            mean_difference /
            difference_sd
        )

    pairwise_overall_results.append({
        "Topic 1": topic_1,
        "Topic 2": topic_2,
        "N": len(paired_data),
        "Mean Difference":
            mean_difference,
        "t": t_statistic,
        "Raw p-value": raw_p,
        "Cohen dz": cohens_dz
    })

pairwise_overall_results = pd.DataFrame(
    pairwise_overall_results
)
```

```python
reject, adjusted_p, _, _ = multipletests(
    pairwise_overall_results[
        "Raw p-value"
    ],
    alpha=0.05,
    method="holm"
)

pairwise_overall_results[
    "Adjusted p-value"
] = adjusted_p

pairwise_overall_results[
    "Significant"
] = reject

display(
    pairwise_overall_results.round(4)
)
```

```python
wilcoxon_overall_results = []

for topic_1, topic_2 in topic_pairs:

    paired_data = overall_wide[
        [topic_1, topic_2]
    ].dropna()

    differences = (
        paired_data[topic_1] -
        paired_data[topic_2]
    )

    if np.allclose(differences, 0):
        statistic = 0
        raw_p = 1.0
    else:
        statistic, raw_p = (
            stats.wilcoxon(
                paired_data[topic_1],
                paired_data[topic_2]
            )
        )

    wilcoxon_overall_results.append({
        "Topic 1": topic_1,
        "Topic 2": topic_2,
        "N": len(paired_data),
        "Median Difference":
            differences.median(),
        "Wilcoxon Statistic":
            statistic,
        "Raw p-value":
            raw_p
    })

wilcoxon_overall_results = pd.DataFrame(
    wilcoxon_overall_results
)
```

```python
reject, adjusted_p, _, _ = multipletests(
    wilcoxon_overall_results[
        "Raw p-value"
    ],
    alpha=0.05,
    method="holm"
)

wilcoxon_overall_results[
    "Adjusted p-value"
] = adjusted_p

wilcoxon_overall_results[
    "Significant"
] = reject

display(
    wilcoxon_overall_results.round(4)
)
```

```python
def repeated_measures_anova_for_score(
    participant_topic_dataframe,
    score_column
):
    """
    Run repeated-measures ANOVA using one mean
    score per participant per topic.
    """

    relevant_data = (
        participant_topic_dataframe[
            [
                "participant_id",
                "Topic",
                score_column
            ]
        ]
        .dropna()
    )

    topic_counts = (
        relevant_data
        .groupby("participant_id")["Topic"]
        .nunique()
    )

    complete_ids = topic_counts[
        topic_counts == len(topic_order)
    ].index

    relevant_data = relevant_data[
        relevant_data["participant_id"].isin(
            complete_ids
        )
    ]

    if len(complete_ids) < 2:
        return None

    result = AnovaRM(
        data=relevant_data,
        depvar=score_column,
        subject="participant_id",
        within=["Topic"]
    ).fit()

    return result
```

```python
scores_for_topic_tests = [
    "Overall_Kansei_Score",
    "Visual_Kansei_Score",
    "Content_Kansei_Score",
    "Audio_Narrator_Kansei_Score"
]

for score in scores_for_topic_tests:
    print("\n" + "=" * 60)
    print(score)
    print("=" * 60)

    result = repeated_measures_anova_for_score(
        participant_topic_scores,
        score
    )

    if result is None:
        print(
            "Not enough complete participants."
        )
    else:
        print(result)
```

```python
word_test_results = []

for column in kansei_columns:

    participant_word_topic = (
        df.groupby(
            ["participant_id", "Topic"],
            observed=False
        )[column]
        .mean()
        .reset_index()
    )

    word_wide = (
        participant_word_topic
        .pivot(
            index="participant_id",
            columns="Topic",
            values=column
        )
        .reindex(columns=topic_order)
        .dropna()
    )

    if len(word_wide) < 3:
        test_statistic = np.nan
        raw_p = np.nan
    else:
        test_statistic, raw_p = (
            stats.friedmanchisquare(
                word_wide["Language"],
                word_wide["Science"],
                word_wide["History"]
            )
        )

    word_test_results.append({
        "Dimension":
            column_to_dimension[column],
        "Column": column,
        "Kansei Word":
            kansei_labels[column],
        "N Participants":
            len(word_wide),
        "Friedman Statistic":
            test_statistic,
        "Raw p-value":
            raw_p
    })

word_test_results = pd.DataFrame(
    word_test_results
)
```

```python
valid_mask = word_test_results[
    "Raw p-value"
].notna()

adjusted_values = np.full(
    len(word_test_results),
    np.nan
)

significance_values = np.full(
    len(word_test_results),
    False,
    dtype=bool
)

if valid_mask.sum() > 0:

    reject, adjusted_p, _, _ = multipletests(
        word_test_results.loc[
            valid_mask,
            "Raw p-value"
        ],
        alpha=0.05,
        method="fdr_bh"
    )

    adjusted_values[valid_mask] = adjusted_p
    significance_values[valid_mask] = reject

word_test_results[
    "FDR Adjusted p-value"
] = adjusted_values

word_test_results[
    "Significant after FDR"
] = significance_values

word_test_results = (
    word_test_results
    .sort_values(
        "FDR Adjusted p-value"
    )
)

display(word_test_results.round(4))
```

```python
significant_words = (
    word_test_results.loc[
        word_test_results[
            "Significant after FDR"
        ],
        "Column"
    ]
    .tolist()
)

print("Significant Kansei words:")
print(significant_words)
```

```python
word_pairwise_results = []

for column in significant_words:

    participant_word_topic = (
        df.groupby(
            ["participant_id", "Topic"],
            observed=False
        )[column]
        .mean()
        .reset_index()
    )

    word_wide = (
        participant_word_topic
        .pivot(
            index="participant_id",
            columns="Topic",
            values=column
        )
        .reindex(columns=topic_order)
    )

    current_results = []

    for topic_1, topic_2 in topic_pairs:

        paired = word_wide[
            [topic_1, topic_2]
        ].dropna()

        differences = (
            paired[topic_1] -
            paired[topic_2]
        )

        if len(paired) == 0:
            statistic = np.nan
            raw_p = np.nan

        elif np.allclose(differences, 0):
            statistic = 0
            raw_p = 1.0

        else:
            statistic, raw_p = (
                stats.wilcoxon(
                    paired[topic_1],
                    paired[topic_2]
                )
            )

        current_results.append({
            "Column": column,
            "Kansei Word":
                kansei_labels[column],
            "Topic 1": topic_1,
            "Topic 2": topic_2,
            "N": len(paired),
            "Mean Topic 1":
                paired[topic_1].mean(),
            "Mean Topic 2":
                paired[topic_2].mean(),
            "Mean Difference":
                differences.mean(),
            "Wilcoxon Statistic":
                statistic,
            "Raw p-value":
                raw_p
        })

    current_results = pd.DataFrame(
        current_results
    )

    valid_pairwise = current_results[
        "Raw p-value"
    ].notna()

    adjusted = np.full(
        len(current_results),
        np.nan
    )

    rejected = np.full(
        len(current_results),
        False,
        dtype=bool
    )

    if valid_pairwise.sum() > 0:

        reject, adjusted_p, _, _ = (
            multipletests(
                current_results.loc[
                    valid_pairwise,
                    "Raw p-value"
                ],
                alpha=0.05,
                method="holm"
            )
        )

        adjusted[valid_pairwise] = (
            adjusted_p
        )

        rejected[valid_pairwise] = reject

    current_results[
        "Holm Adjusted p-value"
    ] = adjusted

    current_results[
        "Significant"
    ] = rejected

    word_pairwise_results.append(
        current_results
    )
```

```python
if word_pairwise_results:

    word_pairwise_results = pd.concat(
        word_pairwise_results,
        ignore_index=True
    )

    display(
        word_pairwise_results.round(4)
    )

else:
    word_pairwise_results = pd.DataFrame()

    print(
        "No Kansei words were significant "
        "after FDR correction."
    )
```

### 3.2 Relationship Between Kansei Score and Popularity

```python
kansei_score_variables = {
    "Overall_Kansei_Mean":
        "Overall Kansei Score",

    "Visual_Kansei_Mean":
        "Visual Kansei Score",

    "Content_Kansei_Mean":
        "Content Kansei Score",

    "Audio_Narrator_Kansei_Mean":
        "Audio & Narrator Kansei Score"
}
```

```python
missing_kansei_scores = [
    column
    for column in kansei_score_variables
    if column not in video_analysis_complete.columns
]

if missing_kansei_scores:
    raise ValueError(
        "The following Kansei Score columns "
        f"are missing: {missing_kansei_scores}"
    )
```

```python
kansei_ps_results = []

for kansei_column, kansei_name in (
    kansei_score_variables.items()
):

    correlation_data = (
        video_analysis_complete[
            [
                kansei_column,
                "PS"
            ]
        ]
        .dropna()
    )

    n = len(correlation_data)

    if n >= 3:

        pearson_r, pearson_p = (
            stats.pearsonr(
                correlation_data[
                    kansei_column
                ],
                correlation_data["PS"]
            )
        )

        spearman_rho, spearman_p = (
            stats.spearmanr(
                correlation_data[
                    kansei_column
                ],
                correlation_data["PS"]
            )
        )

    else:
        pearson_r = np.nan
        pearson_p = np.nan
        spearman_rho = np.nan
        spearman_p = np.nan

    kansei_ps_results.append({
        "Kansei Measure": kansei_name,
        "Kansei Column": kansei_column,
        "N Videos": n,
        "Pearson r": pearson_r,
        "Pearson p-value": pearson_p,
        "Spearman rho": spearman_rho,
        "Spearman p-value": spearman_p
    })

kansei_ps_correlations = pd.DataFrame(
    kansei_ps_results
)

display(
    kansei_ps_correlations.round(4)
)
```

```python
valid_mask = (
    kansei_ps_correlations[
        "Spearman p-value"
    ].notna()
)

kansei_ps_correlations[
    "Adjusted p-value"
] = np.nan

kansei_ps_correlations[
    "Significant"
] = False

if valid_mask.sum() > 0:

    reject, adjusted_p, _, _ = (
        multipletests(
            kansei_ps_correlations.loc[
                valid_mask,
                "Spearman p-value"
            ],
            alpha=0.05,
            method="holm"
        )
    )

    kansei_ps_correlations.loc[
        valid_mask,
        "Adjusted p-value"
    ] = adjusted_p

    kansei_ps_correlations.loc[
        valid_mask,
        "Significant"
    ] = reject

display(
    kansei_ps_correlations.round(4)
)
```

```python
valid_mask = (
    kansei_ps_correlations[
        "Spearman p-value"
    ].notna()
)

kansei_ps_correlations[
    "Adjusted p-value"
] = np.nan

kansei_ps_correlations[
    "Significant"
] = False

if valid_mask.sum() > 0:

    reject, adjusted_p, _, _ = (
        multipletests(
            kansei_ps_correlations.loc[
                valid_mask,
                "Spearman p-value"
            ],
            alpha=0.05,
            method="holm"
        )
    )

    kansei_ps_correlations.loc[
        valid_mask,
        "Adjusted p-value"
    ] = adjusted_p

    kansei_ps_correlations.loc[
        valid_mask,
        "Significant"
    ] = reject

display(
    kansei_ps_correlations.round(4)
)
```

```python
for kansei_column, kansei_name in (
    kansei_score_variables.items()
):

    plot_data = (
        video_analysis_complete[
            [
                kansei_column,
                "PS"
            ]
        ]
        .dropna()
    )

    if len(plot_data) < 3:
        continue

    x = plot_data[kansei_column]
    y = plot_data["PS"]

    slope, intercept, _, _, _ = (
        stats.linregress(x, y)
    )

    x_line = np.linspace(
        x.min(),
        x.max(),
        100
    )

    y_line = intercept + slope * x_line

    plt.figure(figsize=(8, 5))

    plt.scatter(
        x,
        y,
        s=65
    )

    plt.plot(
        x_line,
        y_line,
        linestyle="--"
    )

    plt.axvline(
        x=0,
        linestyle=":",
        linewidth=1
    )

    plt.axhline(
        y=0,
        linestyle=":",
        linewidth=1
    )

    plt.xlabel(kansei_name)
    plt.ylabel("Popularity Score (PS)")

    plt.title(
        f"{kansei_name} and Popularity Score"
    )

    plt.tight_layout()
    plt.show()
```

```python
popular_unpopular_results = []

for score_column, score_name in (
    comparison_variables.items()
):

    popular_values = (
        video_analysis_complete.loc[
            video_analysis_complete[
                "Popularity_Group"
            ] == "Popular",
            score_column
        ]
        .dropna()
    )

    unpopular_values = (
        video_analysis_complete.loc[
            video_analysis_complete[
                "Popularity_Group"
            ] == "Unpopular",
            score_column
        ]
        .dropna()
    )

    if (
        len(popular_values) > 0
        and len(unpopular_values) > 0
    ):

        u_statistic, raw_p = (
            stats.mannwhitneyu(
                popular_values,
                unpopular_values,
                alternative="two-sided",
                method="auto"
            )
        )

        rank_biserial = (
            (
                2 * u_statistic
            )
            /
            (
                len(popular_values)
                * len(unpopular_values)
            )
            - 1
        )

        pooled_denominator = (
            len(popular_values)
            + len(unpopular_values)
            - 2
        )

        if pooled_denominator > 0:

            pooled_variance = (
                (
                    len(popular_values) - 1
                )
                * popular_values.var(ddof=1)
                +
                (
                    len(unpopular_values) - 1
                )
                * unpopular_values.var(ddof=1)
            ) / pooled_denominator

            pooled_sd = np.sqrt(
                pooled_variance
            )

            if pooled_sd > 0:
                cohens_d = (
                    popular_values.mean()
                    - unpopular_values.mean()
                ) / pooled_sd
            else:
                cohens_d = np.nan

        else:
            cohens_d = np.nan

    else:
        u_statistic = np.nan
        raw_p = np.nan
        rank_biserial = np.nan
        cohens_d = np.nan

    popular_unpopular_results.append({
        "Kansei Measure":
            score_name,
        "Kansei Column":
            score_column,
        "Popular N":
            len(popular_values),
        "Unpopular N":
            len(unpopular_values),
        "Popular Mean":
            popular_values.mean(),
        "Unpopular Mean":
            unpopular_values.mean(),
        "Mean Difference":
            popular_values.mean()
            - unpopular_values.mean(),
        "Mann-Whitney U":
            u_statistic,
        "Raw p-value":
            raw_p,
        "Rank-Biserial Correlation":
            rank_biserial,
        "Cohen d":
            cohens_d
    })

popular_unpopular_results = (
    pd.DataFrame(
        popular_unpopular_results
    )
)
```

```python
valid_mask = (
    popular_unpopular_results[
        "Raw p-value"
    ].notna()
)

popular_unpopular_results[
    "Adjusted p-value"
] = np.nan

popular_unpopular_results[
    "Significant"
] = False

if valid_mask.sum() > 0:

    reject, adjusted_p, _, _ = (
        multipletests(
            popular_unpopular_results.loc[
                valid_mask,
                "Raw p-value"
            ],
            alpha=0.05,
            method="holm"
        )
    )

    popular_unpopular_results.loc[
        valid_mask,
        "Adjusted p-value"
    ] = adjusted_p

    popular_unpopular_results.loc[
        valid_mask,
        "Significant"
    ] = reject

display(
    popular_unpopular_results.round(4)
)
```

```python
# Calculate video-level average scores
video_relationship_data = (
    df.groupby(
        [
            "video_id_count",
            "group",
            "video_id",
            "video_title",
            "Topic"
        ],
        observed=True,   # Changed from False to True
        dropna=False
    )[kansei_columns + ["overall_rating"]]
    .mean()
    .reset_index()
)


# Count the number of evaluations and participants per video
video_response_counts = (
    df.groupby(
        "video_id_count",
        observed=True
    )
    .agg(
        Number_of_Evaluations=(
            "participant_id",
            "count"
        ),
        Number_of_Participants=(
            "participant_id",
            "nunique"
        )
    )
    .reset_index()
)


# Merge the counts with the mean scores
video_relationship_data = (
    video_relationship_data.merge(
        video_response_counts,
        on="video_id_count",
        how="left"
    )
)


# Sort according to video number
video_relationship_data["video_number"] = (
    video_relationship_data["video_id_count"]
    .str.extract(r"(\d+)", expand=False)
    .astype(int)
)

video_relationship_data = (
    video_relationship_data
    .sort_values("video_number")
    .drop(columns="video_number")
    .reset_index(drop=True)
)


print(
    "Number of rows:",
    len(video_relationship_data)
)

print(
    "Number of unique videos:",
    video_relationship_data["video_id_count"].nunique()
)

display(
    video_relationship_data[
        [
            "video_id_count",
            "group",
            "video_id",
            "video_title",
            "Topic",
            "Number_of_Evaluations",
            "Number_of_Participants",
            "overall_rating"
        ]
        + kansei_columns
    ].round(3)
)
```

### 4.1 Pearson Correlation

```python
overall_correlation_results = []


for column in kansei_columns:

    correlation_data = (
        video_relationship_data[
            [column, "overall_rating"]
        ]
        .dropna()
    )

    # Pearson correlation requires variation in both variables
    if (
        len(correlation_data) >= 3
        and correlation_data[column].nunique() > 1
        and correlation_data["overall_rating"].nunique() > 1
    ):
        r_value, p_value = stats.pearsonr(
            correlation_data[column],
            correlation_data["overall_rating"]
        )
    else:
        r_value = np.nan
        p_value = np.nan

    overall_correlation_results.append({
        "Kansei Column": column,
        "Kansei Word":
            kansei_readable_names.get(
                column,
                column
            ),
        "N Videos": len(correlation_data),
        "Pearson r": r_value,
        "Raw p-value": p_value,
        "R-squared": (
            r_value ** 2
            if pd.notna(r_value)
            else np.nan
        )
    })


overall_correlation_results = pd.DataFrame(
    overall_correlation_results
)


# Holm correction for the 15 correlation tests
valid_p_mask = (
    overall_correlation_results[
        "Raw p-value"
    ].notna()
)

overall_correlation_results[
    "Adjusted p-value"
] = np.nan

overall_correlation_results[
    "Significant"
] = False

if valid_p_mask.sum() > 0:

    reject, adjusted_p, _, _ = multipletests(
        overall_correlation_results.loc[
            valid_p_mask,
            "Raw p-value"
        ],
        alpha=0.05,
        method="holm"
    )

    overall_correlation_results.loc[
        valid_p_mask,
        "Adjusted p-value"
    ] = adjusted_p

    overall_correlation_results.loc[
        valid_p_mask,
        "Significant"
    ] = reject


# Absolute correlation for ranking
overall_correlation_results[
    "Absolute r"
] = (
    overall_correlation_results[
        "Pearson r"
    ].abs()
)


# Sort from strongest to weakest relationship
overall_correlation_results = (
    overall_correlation_results
    .sort_values(
        "Absolute r",
        ascending=False
    )
    .reset_index(drop=True)
)


overall_correlation_results[
    "Rank"
] = (
    np.arange(
        1,
        len(overall_correlation_results) + 1
    )
)


display(
    overall_correlation_results[
        [
            "Rank",
            "Kansei Word",
            "N Videos",
            "Pearson r",
            "R-squared",
            "Raw p-value",
            "Adjusted p-value",
            "Significant"
        ]
    ]
    .round(4)
)
```

```python
# Add an interpretation of correlation strength
def interpret_correlation(r_value):

    if pd.isna(r_value):
        return "Not available"

    absolute_r = abs(r_value)

    if absolute_r < 0.10:
        strength = "Negligible"
    elif absolute_r < 0.30:
        strength = "Weak"
    elif absolute_r < 0.50:
        strength = "Moderate"
    elif absolute_r < 0.70:
        strength = "Strong"
    else:
        strength = "Very strong"

    direction = (
        "positive"
        if r_value > 0
        else "negative"
        if r_value < 0
        else "none"
    )

    return f"{strength} {direction}"


overall_correlation_results[
    "Interpretation"
] = (
    overall_correlation_results[
        "Pearson r"
    ]
    .apply(interpret_correlation)
)


display(
    overall_correlation_results[
        [
            "Rank",
            "Kansei Word",
            "Pearson r",
            "Adjusted p-value",
            "Significant",
            "Interpretation"
        ]
    ]
    .round(4)
)
```

```python
# Bar chart of Pearson correlations
correlation_plot_data = (
    overall_correlation_results
    .sort_values(
        "Pearson r",
        ascending=True
    )
)


plt.figure(figsize=(10, 8))

plt.barh(
    correlation_plot_data[
        "Kansei Word"
    ],
    correlation_plot_data[
        "Pearson r"
    ]
)

plt.axvline(
    x=0,
    linestyle="--",
    linewidth=1
)

plt.xlabel(
    "Pearson Correlation with Overall Rating"
)

plt.ylabel(
    "Kansei Adjective"
)

plt.title(
    "Relationship Between Kansei Adjectives "
    "and Overall Video Evaluation"
)

plt.xlim(-1, 1)

plt.tight_layout()
plt.show()
```

```python
topic_correlation_results = []


for topic_name in topic_order:

    topic_video_data = (
        video_relationship_data[
            video_relationship_data[
                "Topic"
            ] == topic_name
        ]
        .copy()
    )


    for column in kansei_columns:

        correlation_data = (
            topic_video_data[
                [column, "overall_rating"]
            ]
            .dropna()
        )

        if (
            len(correlation_data) >= 3
            and correlation_data[column].nunique() > 1
            and correlation_data[
                "overall_rating"
            ].nunique() > 1
        ):
            r_value, p_value = stats.pearsonr(
                correlation_data[column],
                correlation_data[
                    "overall_rating"
                ]
            )
        else:
            r_value = np.nan
            p_value = np.nan


        topic_correlation_results.append({
            "Topic": topic_name,
            "Kansei Column": column,
            "Kansei Word":
                kansei_readable_names.get(
                    column,
                    column
                ),
            "N Videos": len(correlation_data),
            "Pearson r": r_value,
            "Raw p-value": p_value,
            "R-squared": (
                r_value ** 2
                if pd.notna(r_value)
                else np.nan
            )
        })


topic_correlation_results = pd.DataFrame(
    topic_correlation_results
)
```

```python
# Holm correction separately within each topic
topic_correlation_results[
    "Adjusted p-value"
] = np.nan

topic_correlation_results[
    "Significant"
] = False


for topic_name in topic_order:

    topic_mask = (
        topic_correlation_results[
            "Topic"
        ] == topic_name
    )

    valid_mask = (
        topic_mask
        & topic_correlation_results[
            "Raw p-value"
        ].notna()
    )

    if valid_mask.sum() > 0:

        reject, adjusted_p, _, _ = multipletests(
            topic_correlation_results.loc[
                valid_mask,
                "Raw p-value"
            ],
            alpha=0.05,
            method="holm"
        )

        topic_correlation_results.loc[
            valid_mask,
            "Adjusted p-value"
        ] = adjusted_p

        topic_correlation_results.loc[
            valid_mask,
            "Significant"
        ] = reject


topic_correlation_results[
    "Absolute r"
] = (
    topic_correlation_results[
        "Pearson r"
    ].abs()
)


topic_correlation_results[
    "Interpretation"
] = (
    topic_correlation_results[
        "Pearson r"
    ]
    .apply(interpret_correlation)
)


# Rank within each topic
topic_correlation_results[
    "Rank Within Topic"
] = (
    topic_correlation_results
    .groupby("Topic", observed=False)[
        "Absolute r"
    ]
    .rank(
        method="dense",
        ascending=False
    )
)


topic_correlation_results = (
    topic_correlation_results
    .sort_values(
        [
            "Topic",
            "Absolute r"
        ],
        ascending=[
            True,
            False
        ]
    )
    .reset_index(drop=True)
)


display(
    topic_correlation_results[
        [
            "Topic",
            "Rank Within Topic",
            "Kansei Word",
            "N Videos",
            "Pearson r",
            "R-squared",
            "Raw p-value",
            "Adjusted p-value",
            "Significant",
            "Interpretation"
        ]
    ]
    .round(4)
)
```

## 5. Contribution of Kansei Words to Overall Evaluation

```python
# Prepare complete video-level data
regression_columns = (
    kansei_columns
    + ["overall_rating"]
)

overall_regression_data = (
    video_relationship_data[
        [
            "video_id_count",
            "Topic"
        ]
        + regression_columns
    ]
    .dropna()
    .copy()
)


print(
    "Number of videos used:",
    len(overall_regression_data)
)

print(
    "Number of Kansei predictors:",
    len(kansei_columns)
)

print(
    "Residual degrees of freedom before intercept:",
    len(overall_regression_data)
    - len(kansei_columns)
    - 1
)


if len(overall_regression_data) <= len(kansei_columns) + 1:
    print(
        "Warning: There are not enough observations "
        "for ordinary multiple regression."
    )
elif (
    len(overall_regression_data)
    < 5 * len(kansei_columns)
):
    print(
        "Caution: The sample size is small compared with "
        "the number of predictors. Interpret coefficients carefully."
    )
```

### 5.1 Multiple Linear Regression

```python
X_overall_raw = (
    overall_regression_data[
        kansei_columns
    ]
    .copy()
)

y_overall_raw = (
    overall_regression_data[
        "overall_rating"
    ]
    .copy()
)


# Standardize both predictors and outcome
x_scaler_overall = StandardScaler()
y_scaler_overall = StandardScaler()

X_overall_standardized = pd.DataFrame(
    x_scaler_overall.fit_transform(
        X_overall_raw
    ),
    columns=kansei_columns,
    index=X_overall_raw.index
)

y_overall_standardized = pd.Series(
    y_scaler_overall.fit_transform(
        y_overall_raw.to_numpy().reshape(-1, 1)
    ).flatten(),
    index=y_overall_raw.index,
    name="overall_rating_z"
)


display(
    X_overall_standardized
    .head()
    .round(3)
)
```

```python
# Add the intercept
X_overall_ols = sm.add_constant(
    X_overall_standardized
)


# Fit the multiple linear regression
overall_ols_model = sm.OLS(
    y_overall_standardized,
    X_overall_ols
).fit()


print(
    overall_ols_model.summary()
)
```

```python
# Extract the regression results
overall_regression_results = pd.DataFrame({
    "Kansei Column":
        overall_ols_model.params.index,

    "Standardized Beta":
        overall_ols_model.params.values,

    "Standard Error":
        overall_ols_model.bse.values,

    "t-value":
        overall_ols_model.tvalues.values,

    "p-value":
        overall_ols_model.pvalues.values,

    "CI Lower":
        overall_ols_model.conf_int()[0].values,

    "CI Upper":
        overall_ols_model.conf_int()[1].values
})


# Remove the intercept from the Kansei ranking table
overall_regression_results = (
    overall_regression_results[
        overall_regression_results[
            "Kansei Column"
        ] != "const"
    ]
    .copy()
)


overall_regression_results[
    "Kansei Word"
] = (
    overall_regression_results[
        "Kansei Column"
    ]
    .map(kansei_readable_names)
)


overall_regression_results[
    "Absolute Beta"
] = (
    overall_regression_results[
        "Standardized Beta"
    ]
    .abs()
)


overall_regression_results[
    "Significant"
] = (
    overall_regression_results[
        "p-value"
    ] < 0.05
)


overall_regression_results = (
    overall_regression_results
    .sort_values(
        "Absolute Beta",
        ascending=False
    )
    .reset_index(drop=True)
)


overall_regression_results[
    "Rank"
] = (
    np.arange(
        1,
        len(overall_regression_results) + 1
    )
)


display(
    overall_regression_results[
        [
            "Rank",
            "Kansei Word",
            "Standardized Beta",
            "Standard Error",
            "t-value",
            "p-value",
            "CI Lower",
            "CI Upper",
            "Significant"
        ]
    ]
    .round(4)
)
```

### 5.3 Topic-Specific Regression

```python
ridge_alphas = np.logspace(
    -4,
    4,
    200
)

loo = LeaveOneOut()
```

```python
topic_ridge_results = []
topic_model_performance = []
topic_ridge_models = {}


for topic_name in topic_order:

    # Select the videos belonging to one topic
    topic_data = (
        video_relationship_data[
            video_relationship_data["Topic"] == topic_name
        ][
            kansei_columns + ["overall_rating"]
        ]
        .dropna()
        .copy()
    )


    print("=" * 70)
    print(topic_name.upper())
    print("=" * 70)

    print(
        "Number of videos:",
        len(topic_data)
    )


    # --------------------------------------------------------
    # Prepare predictors and outcome
    # --------------------------------------------------------

    X_topic_raw = (
        topic_data[kansei_columns]
        .copy()
    )

    y_topic_raw = (
        topic_data["overall_rating"]
        .copy()
    )


    # Check whether there are enough videos
    if len(topic_data) < 3:

        print(
            f"Skipped {topic_name}: "
            "fewer than three complete videos."
        )

        continue


    # --------------------------------------------------------
    # Standardize predictors and outcome
    # --------------------------------------------------------

    x_scaler_topic = StandardScaler()
    y_scaler_topic = StandardScaler()


    X_topic_z = (
        x_scaler_topic.fit_transform(
            X_topic_raw
        )
    )


    y_topic_z = (
        y_scaler_topic.fit_transform(
            y_topic_raw
            .to_numpy()
            .reshape(-1, 1)
        )
        .flatten()
    )


    # --------------------------------------------------------
    # Select the Ridge alpha using LOOCV
    # --------------------------------------------------------

    ridge_cv_model = RidgeCV(
        alphas=ridge_alphas,
        cv=LeaveOneOut(),
        scoring="neg_mean_squared_error"
    )


    ridge_cv_model.fit(
        X_topic_z,
        y_topic_z
    )


    selected_alpha = (
        ridge_cv_model.alpha_
    )


    print(
        "Selected Ridge alpha:",
        round(selected_alpha, 6)
    )


    # --------------------------------------------------------
    # Fit the final Ridge model using all topic videos
    # --------------------------------------------------------

    final_ridge_model = Ridge(
        alpha=selected_alpha
    )


    final_ridge_model.fit(
        X_topic_z,
        y_topic_z
    )


    # Store the model and scalers
    topic_ridge_models[topic_name] = {
        "model": final_ridge_model,
        "x_scaler": x_scaler_topic,
        "y_scaler": y_scaler_topic,
        "alpha": selected_alpha
    }


    # --------------------------------------------------------
    # Training performance
    # --------------------------------------------------------

    training_predictions_z = (
        final_ridge_model.predict(
            X_topic_z
        )
    )


    training_r2 = r2_score(
        y_topic_z,
        training_predictions_z
    )


    training_mae = mean_absolute_error(
        y_topic_z,
        training_predictions_z
    )


    training_rmse = np.sqrt(
        mean_squared_error(
            y_topic_z,
            training_predictions_z
        )
    )


    # --------------------------------------------------------
    # Manual leave-one-out cross-validation predictions
    # --------------------------------------------------------

    loo = LeaveOneOut()

    loocv_predictions_z = np.zeros(
        len(y_topic_z)
    )


    for train_index, test_index in loo.split(
        X_topic_z
    ):

        # Training and test data for this fold
        X_train = X_topic_z[train_index]
        X_test = X_topic_z[test_index]

        y_train = y_topic_z[train_index]


        # Fit Ridge model to the remaining videos
        fold_model = Ridge(
            alpha=selected_alpha
        )


        fold_model.fit(
            X_train,
            y_train
        )


        # Predict the omitted video
        loocv_predictions_z[test_index] = (
            fold_model.predict(
                X_test
            )
        )


    # Calculate performance after collecting
    # predictions for all omitted videos
    loocv_r2 = r2_score(
        y_topic_z,
        loocv_predictions_z
    )


    loocv_mae = mean_absolute_error(
        y_topic_z,
        loocv_predictions_z
    )


    loocv_rmse = np.sqrt(
        mean_squared_error(
            y_topic_z,
            loocv_predictions_z
        )
    )


    # --------------------------------------------------------
    # Save model performance
    # --------------------------------------------------------

    topic_model_performance.append({
        "Topic": topic_name,
        "N Videos": len(topic_data),
        "Selected Alpha": selected_alpha,

        "Training R-squared":
            training_r2,

        "Training MAE":
            training_mae,

        "Training RMSE":
            training_rmse,

        "LOOCV R-squared":
            loocv_r2,

        "LOOCV MAE":
            loocv_mae,

        "LOOCV RMSE":
            loocv_rmse
    })


    # --------------------------------------------------------
    # Save standardized Ridge coefficients
    # --------------------------------------------------------

    for column, coefficient in zip(
        kansei_columns,
        final_ridge_model.coef_
    ):

        topic_ridge_results.append({
            "Topic":
                topic_name,

            "Kansei Column":
                column,

            "Kansei Word":
                kansei_readable_names.get(
                    column,
                    column
                ),

            "Standardized Ridge Coefficient":
                coefficient,

            "Absolute Coefficient":
                abs(coefficient),

            "Selected Alpha":
                selected_alpha
        })


# Convert the collected results into DataFrames
topic_ridge_results = pd.DataFrame(
    topic_ridge_results
)


topic_model_performance = pd.DataFrame(
    topic_model_performance
)


print("\nTopic-specific Ridge regression completed.")
```

### 5.5 Overall Score-Based Kansei Profile

```python
# Make sure overall_rating is numeric
df["overall_rating"] = pd.to_numeric(
    df["overall_rating"],
    errors="coerce"
)


# Keep only the columns needed for this analysis
overall_score_profile_data = (
    df[
        ["overall_rating"] + kansei_columns
    ]
    .dropna(
        subset=["overall_rating"]
    )
    .copy()
)


# Remove rows where every Kansei score is missing
overall_score_profile_data = (
    overall_score_profile_data[
        ~overall_score_profile_data[
            kansei_columns
        ]
        .isna()
        .all(axis=1)
    ]
    .copy()
)


print(
    "Number of evaluation rows used:",
    len(overall_score_profile_data)
)

print(
    "Overall score values:",
    sorted(
        overall_score_profile_data[
            "overall_rating"
        ]
        .dropna()
        .unique()
    )
)
```

```python
# Number of evaluations for each overall score
overall_score_counts = (
    overall_score_profile_data
    .groupby(
        "overall_rating"
    )
    .size()
    .reset_index(
        name="Number of Evaluations"
    )
    .sort_values(
        "overall_rating"
    )
)


display(
    overall_score_counts
)
```

```python
# Create long-format Kansei table
# Required before ranking

# Mean Kansei score for each overall rating
overall_score_kansei_means = (
    overall_score_profile_data
    .groupby("overall_rating")[kansei_columns]
    .mean()
    .reset_index()
    .sort_values("overall_rating")
)


# Convert from wide format to long format
overall_score_kansei_long = (
    overall_score_kansei_means
    .melt(
        id_vars="overall_rating",
        value_vars=kansei_columns,
        var_name="Kansei Column",
        value_name="Mean Kansei Score"
    )
)


# Add readable Kansei names
overall_score_kansei_long["Kansei Word"] = (
    overall_score_kansei_long["Kansei Column"]
    .map(kansei_readable_names)
)


display(
    overall_score_kansei_long
    .head(20)
    .round(3)
)
```

```python
# Rank Kansei words within each overall score
# Handles tied mean scores
overall_score_kansei_ranking = (
    overall_score_kansei_long
    .sort_values(
        [
            "overall_rating",
            "Mean Kansei Score",
            "Kansei Word"
        ],
        ascending=[
            True,
            False,
            True
        ]
    )
    .reset_index(drop=True)
)


# Assign a unique position within each overall score
# cumcount starts at 0, so add 1
overall_score_kansei_ranking[
    "Rank Within Overall Score"
] = (
    overall_score_kansei_ranking
    .groupby("overall_rating")
    .cumcount()
    + 1
)


display(
    overall_score_kansei_ranking[
        [
            "overall_rating",
            "Rank Within Overall Score",
            "Kansei Word",
            "Mean Kansei Score"
        ]
    ]
    .round(3)
)
```

### 5.6 Regression Assumptions

```python
# Independent variables
X = df[kansei_columns].copy()

# Add intercept
X = sm.add_constant(X)

# Calculate VIF
vif_results = pd.DataFrame({
    "Kansei Word": X.columns,
    "VIF": [
        variance_inflation_factor(X.values, i)
        for i in range(X.shape[1])
    ]
})

# Remove intercept
vif_results = vif_results[vif_results["Kansei Word"] != "const"]

# Interpretation
vif_results["Interpretation"] = pd.cut(
    vif_results["VIF"],
    bins=[0,5,10,float("inf")],
    labels=[
        "Low",
        "Moderate",
        "High"
    ]
)

display(vif_results.sort_values("VIF", ascending=False))
```

```python
# Dependent variable
y = df["overall_rating"]

# Regression model
model = sm.OLS(y, X).fit()

# Residuals
residuals = model.resid

# Shapiro-Wilk test
shapiro_stat, shapiro_p = shapiro(residuals)

normality_results = pd.DataFrame({
    "Statistic":[shapiro_stat],
    "p-value":[shapiro_p],
    "Normal Residuals":[shapiro_p > 0.05]
})

display(normality_results)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))

plt.scatter(
    model.fittedvalues,
    residuals,
    alpha=0.7
)

plt.axhline(0, color="red", linestyle="--")

plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.title("Residual Plot")

plt.show()
```

### 5.7 Feature Importance Ranking

```python
X_std = StandardScaler().fit_transform(df[kansei_columns])

X_std = pd.DataFrame(
    X_std,
    columns=kansei_columns
)

X_std = sm.add_constant(X_std)

y = df["overall_rating"]

standardized_model = sm.OLS(y, X_std).fit()
```

```python
feature_importance_ranking = (
    pd.DataFrame({
        "Kansei Word": standardized_model.params.index,
        "Standardized Beta": standardized_model.params.values,
        "p-value": standardized_model.pvalues.values
    })
    .query("`Kansei Word` != 'const'")
)

feature_importance_ranking["Absolute Beta"] = (
    feature_importance_ranking["Standardized Beta"].abs()
)

feature_importance_ranking = (
    feature_importance_ranking
    .sort_values("Absolute Beta", ascending=False)
    .reset_index(drop=True)
)

feature_importance_ranking.insert(
    0,
    "Rank",
    range(1, len(feature_importance_ranking)+1)
)

feature_importance_ranking["Significant"] = (
    feature_importance_ranking["p-value"] < 0.05
)

display(feature_importance_ranking)
```

```python
plt.figure(figsize=(9,6))

plot_data = feature_importance_ranking.sort_values("Standardized Beta")

plt.barh(
    plot_data["Kansei Word"],
    plot_data["Standardized Beta"]
)

plt.xlabel("Standardized Beta")
plt.ylabel("Kansei Word")
plt.title("Feature Importance Ranking")

plt.tight_layout()
plt.show()
```

# Analysis Improve

## 1. Cross-classified mixed-effects model

```python
# ============================================================
# Cross-Classified Mixed-Effects Model
# Random effects: Participant + Video
# Outcome: Overall Evaluation
# Fixed effect: Topic
# ============================================================

# Prepare data
crossed_data = df[
    [
        "participant_id",
        "video_id_count",
        "Topic",
        "overall_rating"
    ]
].dropna().copy()

# A single artificial grouping variable is used because
# Participant and Video are crossed, not nested
crossed_data["all_group"] = 1

print("Number of observations:", len(crossed_data))
print(
    "Number of participants:",
    crossed_data["participant_id"].nunique()
)
print(
    "Number of videos:",
    crossed_data["video_id_count"].nunique()
)

# Cross-classified mixed-effects model
crossed_model = smf.mixedlm(
    formula="overall_rating ~ C(Topic)",
    data=crossed_data,
    groups=crossed_data["all_group"],
    re_formula="0",
    vc_formula={
        "Participant": "0 + C(participant_id)",
        "Video": "0 + C(video_id_count)"
    }
)

crossed_result = crossed_model.fit(
    reml=True,
    method="lbfgs"
)

print(crossed_result.summary())
```

```python
# ============================================================
# Science as reference category
# This gives the Science vs History comparison
# ============================================================

science_ref_model = smf.mixedlm(
    formula=(
        "overall_rating ~ "
        "C(Topic, Treatment(reference='Science'))"
    ),
    data=crossed_data,
    groups=crossed_data["all_group"],
    re_formula="0",
    vc_formula={
        "Participant": "0 + C(participant_id)",
        "Video": "0 + C(video_id_count)"
    }
)

science_ref_result = science_ref_model.fit(
    reml=True,
    method="lbfgs"
)

print(science_ref_result.summary())
```

## 2. Reduced Cross-Classified Mixed-Effects Model

```python
# ============================================================
# Outcome: Overall Evaluation
# Predictors: 3 Kansei dimensions + Topic + Group
# Random effects: Participant + Video
# ============================================================

reduced_data = df[
    [
        "participant_id",
        "video_id_count",
        "Topic",
        "group",
        "Visual_Kansei_Score",
        "Content_Kansei_Score",
        "Audio_Narrator_Kansei_Score",
        "overall_rating"
    ]
].dropna().copy()

# Dummy grouping variable for crossed random effects
reduced_data["all_group"] = 1

print("Number of observations:", len(reduced_data))
print(
    "Number of participants:",
    reduced_data["participant_id"].nunique()
)
print(
    "Number of videos:",
    reduced_data["video_id_count"].nunique()
)

reduced_model = smf.mixedlm(
    formula=(
        "overall_rating ~ "
        "Visual_Kansei_Score + "
        "Content_Kansei_Score + "
        "Audio_Narrator_Kansei_Score + "
        "C(Topic) + "
        "C(group)"
    ),
    data=reduced_data,
    groups=reduced_data["all_group"],
    re_formula="0",
    vc_formula={
        "Participant": "0 + C(participant_id)",
        "Video": "0 + C(video_id_count)"
    }
)

reduced_result = reduced_model.fit(
    reml=True,
    method="lbfgs"
)

print(reduced_result.summary())
```

```python
df["Visual_Mean"] = df[
    [
        "attractive_plain",
        "clear_confusing",
        "organized_disorganized",
        "comfortable_uncomfortable",
        "coherent_scattered"
    ]
].mean(axis=1)

df["Content_Mean"] = df[
    [
        "professional_amateur",
        "understandable_confusing",
        "interesting_boring",
        "factual_misleading",
        "convincing_doubtful"
    ]
].mean(axis=1)

df["Audio_Mean"] = df[
    [
        "pleasant_unpleasant",
        "clear_unclear",
        "smooth_abrupt",
        "friendly_unfriendly",
        "motivating_demotivating"
    ]
].mean(axis=1)
```

```python
reduced_data = df[
    [
        "participant_id",
        "video_id_count",
        "Topic",
        "group",
        "Visual_Mean",
        "Content_Mean",
        "Audio_Mean",
        "overall_rating"
    ]
].dropna().copy()

reduced_data["all_group"] = 1

reduced_model = smf.mixedlm(
    formula=(
        "overall_rating ~ "
        "Visual_Mean + "
        "Content_Mean + "
        "Audio_Mean + "
        "C(Topic) + "
        "C(group)"
    ),
    data=reduced_data,
    groups=reduced_data["all_group"],
    re_formula="0",
    vc_formula={
        "Participant": "0 + C(participant_id)",
        "Video": "0 + C(video_id_count)"
    }
)

reduced_result = reduced_model.fit(
    reml=True,
    method="lbfgs"
)

print(reduced_result.summary())
```

```python
# ============================================================
# Try different optimizers
# ============================================================

methods = ["powell", "bfgs", "cg", "lbfgs"]

results = {}

for method in methods:
    print("\n" + "=" * 60)
    print("Optimizer:", method)
    print("=" * 60)

    try:
        result = reduced_model.fit(
            reml=True,
            method=method,
            maxiter=5000,
            disp=False
        )

        results[method] = result

        print("Converged:", result.converged)
        print("Log-likelihood:", result.llf)

    except Exception as e:
        print("Failed:", e)
```

```python
for method, result in results.items():
    if result.converged:
        print("\nCONVERGED MODEL:", method)
        print(result.summary())
```

```python
# ============================================================
# Participant-level data for the 15 Kansei adjective pairs
# ============================================================

participant_kansei = (
    df.groupby("participant_id")[kansei_columns]
    .mean()
    .dropna()
)

print("Number of participants:", len(participant_kansei))
print("Number of Kansei items:", participant_kansei.shape[1])

display(participant_kansei.head().round(3))
```

```python
from factor_analyzer import FactorAnalyzer
from factor_analyzer.factor_analyzer import (
    calculate_kmo,
    calculate_bartlett_sphericity
)
```

```python
# ============================================================
# KMO and Bartlett's Test
# ============================================================

kmo_all, kmo_model = calculate_kmo(participant_kansei)

bartlett_chi2, bartlett_p = (
    calculate_bartlett_sphericity(participant_kansei)
)

print("Overall KMO:", round(kmo_model, 3))
print("Bartlett chi-square:", round(bartlett_chi2, 3))
print("Bartlett p-value:", round(bartlett_p, 5))
```

```python
# ============================================================
# Eigenvalues
# ============================================================

fa_check = FactorAnalyzer(
    n_factors=15,
    rotation=None
)

fa_check.fit(participant_kansei)

eigenvalues, _ = fa_check.get_eigenvalues()

eigenvalue_table = pd.DataFrame({
    "Factor": range(1, len(eigenvalues) + 1),
    "Eigenvalue": eigenvalues
})

display(eigenvalue_table.round(3))
```

```python
# ============================================================
# 3-Factor Exploratory Factor Analysis
# ============================================================

fa_3 = FactorAnalyzer(
    n_factors=3,
    rotation="varimax",
    method="minres"
)

fa_3.fit(participant_kansei)

loadings = pd.DataFrame(
    fa_3.loadings_,
    index=[
        kansei_labels[column]
        for column in kansei_columns
    ],
    columns=[
        "Factor 1",
        "Factor 2",
        "Factor 3"
    ]
)

display(loadings.round(3))
```

```python
communalities = pd.DataFrame({
    "Kansei Word": [
        kansei_labels[column]
        for column in kansei_columns
    ],
    "Communality": fa_3.get_communalities()
})

display(communalities.round(3))
```

```python
# ============================================================
# 2-Factor Exploratory Factor Analysis
# Oblimin rotation allows correlated factors
# ============================================================

fa_2 = FactorAnalyzer(
    n_factors=2,
    rotation="oblimin",
    method="minres"
)

fa_2.fit(participant_kansei)

loadings_2 = pd.DataFrame(
    fa_2.loadings_,
    index=[
        kansei_labels[column]
        for column in kansei_columns
    ],
    columns=[
        "Factor 1",
        "Factor 2"
    ]
)

display(loadings_2.round(3))
```

```python
variance_2 = fa_2.get_factor_variance()

variance_table_2 = pd.DataFrame(
    variance_2,
    index=[
        "SS Loadings",
        "Proportion Variance",
        "Cumulative Variance"
    ],
    columns=[
        "Factor 1",
        "Factor 2"
    ]
)

display(variance_table_2.round(3))
```

```python
print("Factor correlation matrix:")
display(
    pd.DataFrame(
        fa_2.phi_,
        index=["Factor 1", "Factor 2"],
        columns=["Factor 1", "Factor 2"]
    ).round(3)
)
```

```python
# ============================================================
# Create exploratory factor scores for all 372 evaluations
# Factor structure was estimated from participant-level data
# ============================================================

factor_scores = fa_2.transform(
    df[kansei_columns]
)

df["Visual_Content_Factor"] = factor_scores[:, 0]
df["Audio_Narrator_Factor"] = factor_scores[:, 1]

display(
    df[
        [
            "Visual_Content_Factor",
            "Audio_Narrator_Factor"
        ]
    ].describe().round(3)
)
```

```python
# ============================================================
# Cross-Classified Mixed Model with 2 Exploratory Factors
# ============================================================

factor_model_data = df[
    [
        "participant_id",
        "video_id_count",
        "Topic",
        "group",
        "Visual_Content_Factor",
        "Audio_Narrator_Factor",
        "overall_rating"
    ]
].dropna().copy()

factor_model_data["all_group"] = 1

print("Number of observations:", len(factor_model_data))
print(
    "Number of participants:",
    factor_model_data["participant_id"].nunique()
)
print(
    "Number of videos:",
    factor_model_data["video_id_count"].nunique()
)

factor_mixed_model = smf.mixedlm(
    formula=(
        "overall_rating ~ "
        "Visual_Content_Factor + "
        "Audio_Narrator_Factor + "
        "C(Topic) + "
        "C(group)"
    ),
    data=factor_model_data,
    groups=factor_model_data["all_group"],
    re_formula="0",
    vc_formula={
        "Participant": "0 + C(participant_id)",
        "Video": "0 + C(video_id_count)"
    }
)

factor_mixed_result = factor_mixed_model.fit(
    reml=True,
    method="powell",
    maxiter=5000,
    disp=False
)

print(factor_mixed_result.summary())
```
