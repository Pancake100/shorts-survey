# Participant Anaalysis

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("participant_data.csv")

print("Number of participants:", len(df))
display(df.head())
```

```python
def frequency_table(data, column):
    table = data[column].value_counts(dropna=False).reset_index()
    table.columns = [column, "Count"]

    table["Percentage"] = (
        table["Count"] / table["Count"].sum() * 100
    ).round(1)

    return table
```

## Basic Information

### Group

```python
group_table = frequency_table(df, "group")
display(group_table)
```

### Age

```python
age_labels = {
    "18_22": "18–22",
    "23_27": "23–27"
}

df["age_label"] = df["age"].replace(age_labels)

age_order = ["18–22", "23–27"]

age_table = (
    df["age_label"]
    .value_counts()
    .reindex(age_order, fill_value=0)
    .reset_index()
)

age_table.columns = ["Age", "Count"]
age_table["Percentage"] = (
    age_table["Count"] / len(df) * 100
).round(1)

display(age_table)
```

```python
age_counts = df["age_label"].value_counts().reindex(age_order, fill_value=0)

plt.figure(figsize=(6, 4))
age_counts.plot(kind="bar")

plt.title("Age Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Participants")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Gender

```python
gender_labels = {
    "male": "Male",
    "female": "Female"
}

df["gender_label"] = df["gender"].replace(gender_labels)

gender_table = frequency_table(df, "gender_label")
display(gender_table)
```

```python
gender_counts = df["gender_label"].value_counts()

plt.figure(figsize=(6, 4))
gender_counts.plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Participants")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
country_labels = {
    "taiwan": "Taiwan",
    "indonesia": "Indonesia",
    "china": "China",
    "south_korea": "South Korea",
    "japan": "Japan",
    "other": "Other"
}

df["country_label"] = df["country_region"].replace(country_labels)

country_table = frequency_table(df, "country_label")
display(country_table)
```

```python
country_counts = df["country_label"].value_counts()

plt.figure(figsize=(8, 5))
country_counts.plot(kind="bar")

plt.title("Participants by Country or Region")
plt.xlabel("Country or Region")
plt.ylabel("Number of Participants")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## Short-Video Usage

### Watching Frequency

```python
frequency_labels = {
    "multiple_times_a_day": "Multiple Times a Day",
    "once_a_day": "Once a Day",
    "several_times_a_week": "Several Times a Week",
    "once_a_week": "Once a Week",
    "rarely": "Rarely"
}

df["short_video_frequency_label"] = (
    df["short_video_frequency"].replace(frequency_labels)
)

watch_frequency_table = frequency_table(
    df,
    "short_video_frequency_label"
)

display(watch_frequency_table)
```

```python
watch_frequency_counts = (
    df["short_video_frequency_label"].value_counts()
)

plt.figure(figsize=(7, 4))
watch_frequency_counts.plot(kind="bar")

plt.title("Frequency of Watching Short Videos")
plt.xlabel("Watching Frequency")
plt.ylabel("Number of Participants")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
session_labels = {
    "less_than_10_min": "Less than 10 min",
    "10_30_min": "10–30 min",
    "30_60_min": "30–60 min",
    "1_2_hours": "1–2 hours",
    "more_than_2_hours": "More than 2 hours"
}

df["session_length_label"] = (
    df["usual_session_length"].replace(session_labels)
)

session_order = [
    "Less than 10 min",
    "10–30 min",
    "30–60 min",
    "1–2 hours",
    "More than 2 hours"
]

session_table = (
    df["session_length_label"]
    .value_counts()
    .reindex(session_order, fill_value=0)
    .reset_index()
)

session_table.columns = ["Session Length", "Count"]
session_table["Percentage"] = (
    session_table["Count"] / len(df) * 100
).round(1)

display(session_table)
```

```python
session_counts = (
    df["session_length_label"]
    .value_counts()
    .reindex(session_order, fill_value=0)
)

plt.figure(figsize=(8, 5))
session_counts.plot(kind="bar")

plt.title("Usual Short-Video Watching Session Length")
plt.xlabel("Session Length")
plt.ylabel("Number of Participants")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

### Platforms

```python
platform_data = (
    df["platforms"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
)

platform_labels = {
    "youtube_shorts": "YouTube Shorts",
    "instagram_reels": "Instagram Reels",
    "facebook_reels": "Facebook Reels",
    "tiktok": "TikTok",
    "other": "Other"
}

platform_data = platform_data.replace(platform_labels)

platform_counts = platform_data.value_counts()

platform_table = platform_counts.reset_index()
platform_table.columns = ["Platform", "Count"]

platform_table["Participant Percentage"] = (
    platform_table["Count"] / len(df) * 100
).round(1)

display(platform_table)
```

```python
plt.figure(figsize=(8, 5))
platform_counts.plot(kind="bar")

plt.title("Short-Video Platforms Used by Participants")
plt.xlabel("Platform")
plt.ylabel("Number of Participants")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

### video Types

```python
video_type_data = (
    df["usual_video_types"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
)

video_type_labels = {
    "educational": "Educational",
    "entertainment": "Entertainment",
    "science": "Science",
    "language_learning": "Language Learning",
    "cooking": "Cooking",
    "travel": "Travel",
    "life_hacks": "Life Hacks",
    "gaming": "Gaming",
    "comedy": "Comedy",
    "news_current_events": "News / Current Events",
    "other": "Other"
}

video_type_data = video_type_data.replace(video_type_labels)

video_type_counts = video_type_data.value_counts()

video_type_table = video_type_counts.reset_index()
video_type_table.columns = ["Video Type", "Count"]

video_type_table["Participant Percentage"] = (
    video_type_table["Count"] / len(df) * 100
).round(1)

display(video_type_table)
```

### Topic Preferences

```python
topic_labels = {
    "language_learning": "Language Learning",
    "science": "Science",
    "history": "History"
}

df["topic_preference_1_label"] = (
    df["topic_preference_1"].replace(topic_labels)
)

topic_1_table = frequency_table(
    df,
    "topic_preference_1_label"
)

display(topic_1_table)
```

```python
topic_order = [
    "Language Learning",
    "Science",
    "History"
]

topic_1_counts = (
    df["topic_preference_1_label"]
    .value_counts()
    .reindex(topic_order, fill_value=0)
)

plt.figure(figsize=(7, 5))
topic_1_counts.plot(kind="bar")

plt.title("Participants' First-Choice Educational Topic")
plt.xlabel("Topic")
plt.ylabel("Number of Participants")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```

```python
for column in [
    "topic_preference_1",
    "topic_preference_2",
    "topic_preference_3"
]:
    df[column + "_label"] = df[column].replace(topic_labels)

topic_ranking_table = pd.DataFrame({
    "First Choice": (
        df["topic_preference_1_label"]
        .value_counts()
        .reindex(topic_order, fill_value=0)
    ),
    "Second Choice": (
        df["topic_preference_2_label"]
        .value_counts()
        .reindex(topic_order, fill_value=0)
    ),
    "Third Choice": (
        df["topic_preference_3_label"]
        .value_counts()
        .reindex(topic_order, fill_value=0)
    )
})

display(topic_ranking_table)
```

```python
topic_ranking_table.plot(
    kind="bar",
    figsize=(9, 6)
)

plt.title("Educational Topic Preference Rankings")
plt.xlabel("Topic")
plt.ylabel("Number of Participants")
plt.xticks(rotation=20, ha="right")
plt.legend(title="Ranking")
plt.tight_layout()
plt.show()
```

### Weighted topic-preference score

```python
topic_scores = {
    "Language Learning": 0,
    "Science": 0,
    "History": 0
}

for topic in topic_scores:
    topic_scores[topic] += (
        df["topic_preference_1_label"] == topic
    ).sum() * 3

    topic_scores[topic] += (
        df["topic_preference_2_label"] == topic
    ).sum() * 2

    topic_scores[topic] += (
        df["topic_preference_3_label"] == topic
    ).sum() * 1

topic_score_table = pd.DataFrame(
    topic_scores.items(),
    columns=["Topic", "Weighted Preference Score"]
).sort_values(
    "Weighted Preference Score",
    ascending=False
)

display(topic_score_table)
```

```python
topic_score_plot = topic_score_table.set_index(
    "Topic"
)["Weighted Preference Score"]

plt.figure(figsize=(7, 5))
topic_score_plot.plot(kind="bar")

plt.title("Weighted Educational Topic Preference")
plt.xlabel("Topic")
plt.ylabel("Weighted Preference Score")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```
